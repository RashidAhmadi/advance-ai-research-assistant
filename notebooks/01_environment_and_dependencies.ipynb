```python
#Project configuration
from pathlib import Path
import sys
import subprocess
import platform
import importlib.metadata as metadata

PROJECT_ROOT = (
    Path.home()
    / "Schreibtisch"
    / "IfaDW"
    / "NLP"
    / "Final_Project"
    / "advance-ai-research-assistant"
)

print("Project:", PROJECT_ROOT)
print("Python executable:", sys.executable)
print("Python version:", sys.version)
print("Platform:", platform.platform())
```

```python
# Audit installed Python packages
required_packages = [
    "torch",
    "torchvision",
    "transformers",
    "spacy",
    "chromadb",
    "sentence-transformers",
    "fastapi",
    "uvicorn",
    "pydantic",
    "pandas",
    "numpy",
    "pymupdf",
    "pypdf",
    "langchain",
    "langgraph",
    "rank-bm25",
    "pytest",
    "python-multipart",
    "ollama"
]

for package in required_packages:

    try:
        version = metadata.version(package)
        print(f"✓ {package}: {version}")

    except metadata.PackageNotFoundError:
        print(f"✗ {package}: Not installed")
```

```python
# Check Python dependency conflicts
result = subprocess.run(
    [sys.executable, "-m", "pip", "check"],
    capture_output=True,
    text=True
)

print("Dependency check:")

print(result.stdout)

if result.stderr:
    print(result.stderr)
```

```python
# Check JavaScript development environment
import shutil
import subprocess

commands = ["node", "npm"]

for command in commands:

    executable = shutil.which(command)

    if executable is None:
        print(f"✗ {command}: Not found")
        continue

    result = subprocess.run(
        [executable, "--version"],
        capture_output=True,
        text=True
    )

    if result.returncode == 0:
        print(f"✓ {command}: {result.stdout.strip()}")
    else:
        print(f"✗ {command}: {result.stderr}")
```

```python
# Check CUDA and GPU memory
import torch

if torch.cuda.is_available():

    gpu = torch.cuda.get_device_properties(0)

    print("GPU:", gpu.name)
    print("CUDA:", torch.version.cuda)

    print(
        "Total GPU memory:",
        round(gpu.total_memory / 1024**3, 2),
        "GB"
    )

    print(
        "Available GPU memory:",
        round(
            torch.cuda.mem_get_info()[0] / 1024**3,
            2
        ),
        "GB"
    )

else:
    print("CUDA is unavailable.")
```
