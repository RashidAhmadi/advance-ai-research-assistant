AI providers.

### 1.2 Design Principles

- Evidence-based answers with source citations.

- Efficient resource utilization.
- Reproducible experiments.
- Automated testing and measurable performance.
- Maintainable and well-documented code.


## 3. System Architecture

### 3.1 Architectural Overview

Advance AI Research Assistant is designed as a modular, provider-independent
AI research platform. It enables users to upload scientific PDF documents,
process their contents, search across documents, and receive AI-generated
answers supported by document references and page-level citations.

The system follows a modular architecture consisting of the following layers:

1. **Frontend Layer**
   - Next.js
   - JavaScript (not TypeScript)
   - CSS / CSS Modules
   - Document upload and management
   - Chat interface and search results

2. **Backend API Layer**
   - Python
   - FastAPI
   - Pydantic
   - Authentication and configuration management
   - API endpoints for documents, retrieval and conversations

3. **Document Processing Layer**
   - PDF extraction using PyMuPDF and pypdf
   - Text cleaning and preprocessing
   - Metadata extraction
   - Document chunking

4. **Embedding Layer**
   - Sentence Transformers
   - Provider-independent embedding interface
   - CPU and GPU execution options

5. **Retrieval Layer**
   - ChromaDB as the primary vector database
   - Optional FAISS support
   - Semantic search
   - BM25 keyword retrieval
   - Hybrid retrieval

6. **Reranking Layer**
   - Cross-Encoder models
   - Relevance-based result reranking

7. **LLM and Generation Layer**
   - Ollama for local LLM execution
   - Optional Gemini, Groq and OpenAI providers
   - User-configurable API keys
   - Provider-independent LLM interface

8. **Evaluation Layer**
   - Retrieval quality evaluation
   - Answer quality evaluation
   - Citation verification
   - Performance and latency benchmarks

### 3.2 Design Principles

- Modular and maintainable architecture
- Provider-independent AI components
- Local-first execution with optional cloud services
- Hybrid retrieval for improved search quality
- Source-grounded answers with citations
- Efficient resource utilization
- Testability and reproducibility
- Extensibility for future AI providers and vector databases

### 4.3 Main Data Flow

1. **Document Upload:** Users upload scientific PDF documents through the frontend.
2. **Document Processing:** The backend extracts text, cleans content and creates document chunks.
3. **Embedding Generation:** Text chunks are transformed into numerical vector representations.
4. **Data Storage:** Vectors, metadata and keyword indexes are stored in their respective databases.
5. **Hybrid Retrieval:** Semantic search and BM25 keyword retrieval identify relevant passages.
6. **Reranking:** A Cross-Encoder evaluates and reorders retrieved passages.
7. **LLM Generation:** The selected context is sent to the configured local or cloud LLM provider.
8. **Citation Validation:** The system validates document references and page numbers.
9. **Response Delivery:** The frontend displays the generated answer with supporting sources.

### 4.4 Important Architectural Decisions

- **ChromaDB:** Primary vector database.
- **FAISS:** Optional alternative vector indexing implementation.
- **BM25:** Keyword-based retrieval for hybrid search.
- **Cross-Encoder:** Improves the relevance ordering of retrieved passages.
- **Ollama:** Default local LLM option.
- **Gemini, Groq and OpenAI:** Optional cloud providers using user-supplied API keys.
- **FastAPI:** Central backend interface.
- **Provider-independent interfaces:** Embedding and LLM providers can be changed without redesigning the entire application.
- **Citation support:** Generated answers must be linked to original document passages and page numbers.

**Design principle:** The system should be modular, resource-efficient, extensible and capable of running locally without requiring a cloud API key.

## 5.2 Backend Module Responsibilities

The backend is designed as a modular Python application. Each module has a clearly defined responsibility and communicates with other modules through well-defined interfaces.

The goal is to maintain separation of concerns, reduce dependencies between components, and make the system easier to test, extend, and maintain.

### 5.2.1 Core Module

Responsible for application-wide configuration and shared infrastructure.

| Component | Responsibility |
|---|---|
| `config.py` | Load application settings and environment variables |
| `logging.py` | Configure application logging |
| `security.py` | Handle API-key protection and security-related utilities |

### 5.2.2 Document Ingestion Module

Responsible for accepting, extracting, and preparing scientific PDF documents.

| Component | Responsibility |
|---|---|
| `pdf_loader.py` | Load PDF documents |
| `text_extractor.py` | Extract text and page information |
| `preprocessing.py` | Clean and normalize extracted text |
| `chunking.py` | Divide documents into manageable text chunks |
| `metadata.py` | Maintain document and page metadata |

**Expected output:** Clean text chunks with document identifiers, page numbers, and metadata.

### 5.2.3 Embedding Module

Responsible for converting text chunks and user queries into numerical vector representations.

| Component | Responsibility |
|---|---|
| `base.py` | Define a provider-independent embedding interface |
| `sentence_transformer.py` | Implement local embedding models |
| `embedding_manager.py` | Select and manage embedding providers |

The embedding interface should allow us to introduce additional embedding providers without changing the retrieval pipeline.

### 5.2.4 Storage Module

Responsible for storing and retrieving document representations and metadata.

| Component | Responsibility |
|---|---|
| `vector_store.py` | Define a common vector-store interface |
| `chroma_store.py` | Implement ChromaDB integration |
| `faiss_store.py` | Provide optional FAISS integration |
| `keyword_store.py` | Support keyword-based retrieval |
| `metadata_store.py` | Manage document metadata |

**Primary vector database:** ChromaDB  
**Optional vector database:** FAISS

### 5.2.5 Retrieval Module

Responsible for finding the most relevant document chunks for a user query.

| Component | Responsibility |
|---|---|
| `semantic_search.py` | Retrieve chunks using vector similarity |
| `keyword_search.py` | Retrieve chunks using BM25 |
| `hybrid_search.py` | Combine semantic and keyword retrieval |
| `reranker.py` | Reorder results using a Cross-Encoder |
| `context_builder.py` | Prepare relevant context for the LLM |

### 5.2.6 LLM Provider Module

Responsible for communicating with different language model providers.

| Component | Responsibility |
|---|---|
| `base.py` | Define a common LLM interface |
| `ollama_provider.py` | Connect to locally hosted Ollama models |
| `gemini_provider.py` | Connect to Google Gemini |
| `groq_provider.py` | Connect to Groq |
| `openai_provider.py` | Connect to OpenAI |
| `llm_manager.py` | Select and manage the configured provider |

Users should be able to choose between local and cloud-based models.

Cloud providers require user-supplied API keys. API keys must be handled securely and must never be stored in source code or exposed to the frontend.

### 5.2.7 RAG Module

Responsible for coordinating retrieval and answer generation.

| Component | Responsibility |
|---|---|
| `pipeline.py` | Coordinate the complete RAG workflow |
| `prompt_templates.py` | Manage prompts for answer generation |
| `citation_manager.py` | Connect generated answers to their source documents and page numbers |

### 5.2.8 API Module

Responsible for exposing backend functionality through REST API endpoints.

| Component | Responsibility |
|---|---|
| `documents.py` | Upload, list, and manage documents |
| `chat.py` | Handle user questions and generated answers |
| `search.py` | Provide document search functionality |
| `settings.py` | Manage model and provider settings |
| `health.py` | Check backend availability and status |

### 5.2.9 Evaluation Module

Responsible for measuring retrieval and answer-generation quality.

| Component | Responsibility |
|---|---|
| `retrieval_metrics.py` | Evaluate retrieval performance |
| `answer_metrics.py` | Evaluate generated answers |
| `evaluation_pipeline.py` | Coordinate evaluation experiments |

### 5.2.10 Services Module

The service layer connects API endpoints with the application logic.

It keeps API routes lightweight and prevents business logic from being duplicated across endpoints.

### 5.2.11 Design Principles

The backend follows these principles:

1. **Separation of concerns:** Each module has a clearly defined responsibility.
2. **Provider independence:** Embedding and LLM providers can be exchanged without redesigning the RAG pipeline.
3. **Loose coupling:** Modules communicate through interfaces rather than depending on specific implementations.
4. **Testability:** Components should be testable independently.
5. **Scalability:** New providers, retrieval methods, and storage technologies can be integrated later.
6. **Security:** API keys and sensitive configuration remain on the backend.
7. **Maintainability:** Reusable application logic is implemented in Python modules rather than being kept exclusively in notebooks.

## 7. Data Flow Architecture

This section describes how data moves through Advance AI Research Assistant.

The application has two primary workflows:

1. Document ingestion: Processing uploaded PDF documents and storing their representations.
2. Question answering: Retrieving relevant information and generating answers supported by document sources.

### 7.1 Workflow A – Document Ingestion

The document ingestion workflow transforms uploaded PDF files into searchable document chunks.

**Step 1 – PDF Upload**

The user uploads a scientific PDF through the Next.js frontend.

**Step 2 – API Validation**

FastAPI validates the uploaded file, checks its format and size, and assigns a unique document identifier.

**Step 3 – PDF Extraction**

PyMuPDF extracts text from the PDF while preserving page information.

**Step 4 – Text Preprocessing**

The extracted text is cleaned and normalized. Unnecessary whitespace and extraction artifacts are handled without removing meaningful scientific information.

**Step 5 – Document Chunking**

The text is divided into smaller chunks. Each chunk retains metadata such as:

- Document ID
- Document name
- Page number or page range
- Chunk ID
- Chunk position
- Additional document metadata

**Step 6 – Embedding Generation**

The embedding model converts each chunk into a numerical vector representation.

**Step 7 – Storage**

The application stores:
- Text chunks
- Vector embeddings
- Document and page metadata

ChromaDB is the primary vector database. FAISS may be supported as an optional alternative.

**Step 8 – Processing Confirmation**

The backend returns the document processing status to the frontend.

### 7.2 Workflow B – Question Answering

The question-answering workflow uses the stored documents to generate context-aware responses.

**Step 1 – User Question**

The user submits a research question through the chat interface.

**Step 2 – Query Embedding**

The selected embedding model converts the question into a vector representation.

**Step 3 – Semantic Retrieval**

The vector database searches for document chunks with similar semantic meaning.

**Step 4 – Keyword Retrieval**

BM25 identifies relevant chunks based on keyword matching.

**Step 5 – Hybrid Retrieval**

The semantic and keyword retrieval results are combined.

Reciprocal Rank Fusion (RRF) can be used to combine rankings from different retrieval methods.

**Step 6 – Cross-Encoder Reranking**

A Cross-Encoder evaluates the relevance of the retrieved chunks against the original question and reorders them.

**Step 7 – Context Construction**

The most relevant chunks are selected and assembled into a context for the language model.

**Step 8 – LLM Generation**

The configured LLM provider generates an answer using the question and retrieved context.

Supported providers:
- Ollama (local)
- Google Gemini
- Groq
- OpenAI

**Step 9 – Citation Processing**

The citation manager connects the generated answer to its supporting document chunks and page references.

The application should distinguish between directly supported statements and information that cannot be verified from the retrieved sources.

**Step 10 – Response Delivery**

FastAPI returns the answer, citations, and source references to the frontend.

The frontend displays the response and allows users to inspect the original sources.

### 7.3 Data Flow Design Principles

1. Preserve document identifiers and page metadata throughout the complete pipeline.
2. Keep ingestion and question answering as separate workflows.
3. Use the same compatible embedding model for document chunks and queries within a vector collection.
4. Keep retrieval methods modular so that semantic, keyword, and hybrid retrieval can be evaluated independently.
5. Apply reranking before constructing the final LLM context.
6. Generate answers from retrieved evidence rather than relying only on the LLM's internal knowledge.
7. Preserve source references so that answers can be traced back to the original PDF pages.
8. Handle extraction failures, empty documents, and unsupported files gracefully.
9. Record processing and retrieval errors for debugging and evaluation.
10. Keep provider credentials and model execution on the backend.

## 8. Database and Storage Architecture

Advance AI Research Assistant uses a modular storage architecture to manage scientific documents, extracted text, embeddings, and application metadata.

The initial implementation uses ChromaDB as the primary vector database and SQLite as the relational metadata database.

FAISS is considered an optional alternative for vector similarity search.

### 8.1 Storage Technology Overview

| Technology | Purpose | Status |
|---|---|---|
| ChromaDB | Store and search vector embeddings and associated chunk data | Primary |
| SQLite | Store document records, processing status, and application metadata | Primary |
| FAISS | Alternative vector similarity search | Optional |
| PostgreSQL | Potential relational database for larger deployments | Future option |

### 8.2 ChromaDB – Vector Storage

ChromaDB stores the vector representations of document chunks and supports semantic similarity search.

Each stored record should contain:

- Unique chunk ID
- Text content
- Embedding vector
- Document ID
- Page number or page range
- Chunk index
- Additional searchable metadata

A conceptual ChromaDB record:

```python
{
    "id": "document_001_chunk_0001",
    "document": "Extracted text from the document chunk",
    "embedding": [0.12, 0.34, 0.56],
    "metadata": {
        "document_id": "document_001",
        "filename": "research_paper.pdf",
        "page_number": 3,
        "chunk_index": 1
    }
}
```

The embedding values above are illustrative, not actual model output.

### 8.3 ChromaDB Collection Design

The initial application will use a collection for document chunks.

Proposed collection name:

`research_documents`

The collection will contain document chunks from multiple uploaded PDFs.

Every chunk will have a unique identifier and metadata linking it to its original document.

The collection should also record the embedding model and relevant configuration so that incompatible embeddings are not accidentally mixed.

### 8.4 SQLite – Relational Metadata Storage

SQLite will manage structured application information.

Proposed database tables:

**1. documents**

Stores information about uploaded PDF documents.

| Field | Description |
|---|---|
| document_id | Unique document identifier |
| filename | Original filename |
| file_path | Internal storage path |
| file_hash | File checksum for duplicate detection |
| file_size | File size |
| upload_date | Upload timestamp |
| processing_status | Pending, processing, completed, or failed |
| page_count | Number of PDF pages |
| error_message | Processing error, if applicable |

**2. document_chunks**

Stores chunk-level metadata and relationships.

| Field | Description |
|---|---|
| chunk_id | Unique chunk identifier |
| document_id | Related document |
| chunk_index | Position of the chunk |
| page_start | First source page |
| page_end | Last source page |
| token_count | Approximate token count |
| chroma_id | Corresponding ChromaDB record identifier |

The actual chunk text and vector embeddings will primarily reside in ChromaDB.

**3. application_settings**

Stores non-secret application configuration, such as:

- Selected LLM provider
- Selected model name
- Embedding model configuration
- Retrieval settings
- Reranking configuration

API keys must not be stored as ordinary plaintext settings. We will design secure credential handling separately.

### 8.5 Document and Chunk Identification

Every uploaded PDF receives a unique document ID.

Every generated text chunk receives a unique chunk ID.

Example:

```text
Document:
document_001

Chunks:
document_001_chunk_0001
document_001_chunk_0002
document_001_chunk_0003
```

These identifiers allow the application to:

- Retrieve chunks belonging to a particular document.
- Connect search results to their original PDFs.
- Generate accurate page-level citations.
- Delete all chunks belonging to a document.
- Track document-processing status.

Page numbers displayed to users should follow the PDF's one-based page numbering.

### 8.6 Storage Workflow

During document ingestion:

1. Register the uploaded document in SQLite.
2. Extract and preprocess the PDF text.
3. Generate text chunks and their metadata.
4. Generate embeddings.
5. Store chunk records and embeddings in ChromaDB.
6. Store chunk metadata and ChromaDB identifiers in SQLite.
7. Update the document processing status.

During question answering:

1. Generate the query embedding.
2. Search ChromaDB for semantically relevant chunks.
3. Retrieve keyword results using BM25.
4. Combine and rerank the retrieved results.
5. Retrieve source metadata where necessary.
6. Return answers with document and page references.

### 8.7 Data Consistency and Persistence

SQLite and ChromaDB are separate storage systems. Therefore, the application must handle possible failures between the two.

For example, a document could be registered in SQLite but fail during vector insertion.

The application should:

- Track ingestion status.
- Handle partial failures.
- Avoid duplicate chunk identifiers.
- Support document deletion from both storage systems.
- Provide a mechanism to detect and reconcile inconsistent records.
- Preserve original PDF files until deletion is explicitly requested.

SQLite and ChromaDB do not share a single transaction. Consistency must therefore be managed by the application.

### 8.8 Persistence and Backup

The application should use persistent storage directories rather than temporary in-memory databases.

Proposed storage locations:

```text
data/
├── uploads/
│   └── original PDF files
├── processed/
│   └── extracted and processed data
├── chromadb/
│   └── persistent vector database
└── metadata/
    └── application.db
```

The application should support backups of:

- Original uploaded PDFs.
- SQLite database.
- ChromaDB persistent data.
- Relevant configuration and metadata.

### 8.9 Storage Design Principles

1. Separate vector storage from relational metadata.
2. Use ChromaDB as the primary vector database.
3. Keep FAISS optional.
4. Assign stable identifiers to documents and chunks.
5. Preserve source page information for academic citations.
6. Track the embedding model used for each vector collection.
7. Handle partial failures across storage systems.
8. Support document updates and deletion.
9. Use persistent storage and regular backups.
10. Keep the storage layer independent of the retrieval and RAG implementation.

## 9. LLM and Embedding Provider Architecture

Advance AI Research Assistant will support multiple embedding and Large Language Model (LLM) providers.

The architecture uses provider-independent interfaces so that users can change providers without modifying the core RAG pipeline.

Two independent provider systems will be implemented:

1. Embedding Provider Architecture
2. LLM Provider Architecture

These systems have different responsibilities and must not be tightly coupled.

### 9.1 Why Provider Independence?

Different users have different hardware, privacy requirements, budgets, and model preferences.

For example:

- Some users prefer local models because they do not want to send research documents to external services.
- Some users prefer cloud models because they provide access to powerful language models without requiring local GPU resources.
- Some users want to compare different models for research and evaluation.
- Some users may have limited hardware resources.

Therefore, the application should allow users to configure their preferred providers.

### 9.2 LLM Provider Overview

The initial implementation will support the following providers.

| Provider | Execution | API Key Required | Initial Priority |
|---|---|---|---|
| Ollama | Local | No cloud API key | Primary |
| Google Gemini | Cloud | Yes | Supported |
| Groq | Cloud | Yes | Supported |
| OpenAI | Cloud | Yes | Supported |

Ollama will be the primary local provider.

Cloud providers will be optional. Users should be able to use the application without configuring any cloud API keys.

### 9.3 Provider-Independent LLM Interface

A common interface will define how the RAG pipeline communicates with language models.

Proposed interface:

`backend/app/llm/base.py`

Responsibilities:

- Define a common generation method.
- Define a common streaming method where supported.
- Standardize model configuration.
- Standardize response structures.
- Handle provider-independent generation parameters.
- Define common exception types.

Conceptual interface:

```python
from abc import ABC, abstractmethod
from typing import AsyncIterator


class BaseLLMProvider(ABC):

    @abstractmethod
    async def generate(
        self,
        prompt: str,
        system_prompt: str | None = None,
        temperature: float = 0.2,
        max_tokens: int = 2048,
    ) -> str:
        """Generate a response from the configured LLM."""
        raise NotImplementedError

    @abstractmethod
    async def stream(
        self,
        prompt: str,
        system_prompt: str | None = None,
    ) -> AsyncIterator[str]:
        """Stream generated response chunks."""
        raise NotImplementedError
```

This is a conceptual interface. The final implementation may extend it with structured responses, token usage, timeouts, and provider-specific capabilities.

### 9.4 LLM Provider Implementations

Each provider will implement the common interface.

| Module | Responsibility |
|---|---|
| `ollama_provider.py` | Communicate with local Ollama models |
| `gemini_provider.py` | Communicate with Google Gemini |
| `groq_provider.py` | Communicate with Groq |
| `openai_provider.py` | Communicate with OpenAI |
| `llm_manager.py` | Select and manage the configured provider |

The RAG pipeline will communicate with the LLM manager rather than calling provider SDKs directly.

### 9.5 LLM Configuration

Users should be able to configure:

- Provider
- Model name
- API key for cloud providers
- Temperature
- Maximum output tokens
- Request timeout
- Streaming preference

Example configuration:

```python
llm_config = {
    "provider": "ollama",
    "model": "your-local-model",
    "temperature": 0.2,
    "max_tokens": 2048,
    "stream": True
}
```

This is an illustrative configuration. The actual model name will be selected after checking the models available in Ollama.

### 9.6 Embedding Provider Architecture

Embedding models convert text into numerical vector representations.

The application must use a separate embedding interface because embedding generation is fundamentally different from text generation.

The initial embedding implementation will support:

- Local Sentence Transformers models.
- Additional embedding providers in the future.

The embedding architecture should allow future cloud embedding integrations without changing the retrieval pipeline.

Proposed modules:

| Module | Responsibility |
|---|---|
| `embeddings/base.py` | Define the common embedding interface |
| `embeddings/sentence_transformer.py` | Implement local Sentence Transformers |
| `embeddings/embedding_manager.py` | Select and manage embedding models |

### 9.7 Embedding Interface

Conceptual interface:

```python
from abc import ABC, abstractmethod


class BaseEmbeddingProvider(ABC):

    @abstractmethod
    def embed_documents(
        self,
        texts: list[str]
    ) -> list[list[float]]:
        """Generate embeddings for document chunks."""
        raise NotImplementedError

    @abstractmethod
    def embed_query(
        self,
        text: str
    ) -> list[float]:
        """Generate an embedding for a user query."""
        raise NotImplementedError
```

The separate methods allow providers to implement different document and query encoding behavior when required.

### 9.8 Embedding Model Consistency

An important requirement is that document embeddings and query embeddings must be compatible.

For example:

1. Scientific PDFs are processed using a selected embedding model.
2. Their chunks are converted into vectors.
3. The vectors are stored in ChromaDB.
4. User questions are converted into vectors using a compatible model.
5. Similarity search compares the query vector with stored document vectors.

The application must track:

- Embedding provider
- Model name and revision
- Vector dimension
- Distance metric
- Relevant model configuration

Changing the embedding model may require rebuilding the vector collection and regenerating document embeddings.

The application must not silently mix incompatible embedding spaces.

### 9.9 LLM and Embedding Independence

The LLM provider and embedding provider must be configurable independently.

For example, a user could select:

| Embedding Provider | LLM Provider |
|---|---|
| Sentence Transformers (local) | Ollama (local) |
| Sentence Transformers (local) | Gemini |
| Sentence Transformers (local) | Groq |
| Sentence Transformers (local) | OpenAI |

This separation allows users to keep document embedding local while using a cloud LLM for answer generation.

### 9.10 API Key Management and Security

Cloud API keys must be handled securely.

The architecture should follow these principles:

1. API keys must never be hardcoded in Python or JavaScript source code.
2. API keys must never be returned to the frontend.
3. Cloud API keys must be sent to the backend over HTTPS in production.
4. Credentials must not be stored as ordinary plaintext application settings.
5. Environment variables or a suitable secure credential-storage mechanism should be used.
6. Logs must never contain API keys.
7. Users should be able to update or remove their configured credentials.

For an initial single-user local application, environment variables can provide a simple starting point.

For a multi-user deployment, credentials should be stored securely and isolated per user.

### 9.11 Provider Selection Workflow

When a user selects a provider:

1. The frontend submits the selected provider and model configuration.
2. FastAPI validates the configuration.
3. The backend checks provider availability and credentials.
4. The LLM manager initializes the selected provider.
5. The RAG pipeline uses the common LLM interface.
6. The generated answer is returned through the API.

The same principle applies to embedding provider selection, with an additional compatibility check against the active vector collection.

### 9.12 Design Principles

1. Keep LLM and embedding interfaces separate.
2. Use abstract base classes for common provider contracts.
3. Keep provider SDKs isolated inside their implementations.
4. Make local inference possible without cloud API keys.
5. Allow independent configuration of embedding and LLM providers.
6. Validate model availability and provider credentials.
7. Track embedding model information with vector collections.
8. Handle provider-specific errors consistently.
9. Protect user credentials and sensitive configuration.
10. Allow new providers to be added without redesigning the RAG pipeline.

### 10.9 Future Authentication and Multi-user Support

The initial implementation will focus on a local single-user application.

However, the architecture should allow future integration of:

- User registration and login.
- Password hashing.
- Session management or token-based authentication.
- Role-based access control.
- User-specific documents and collections.
- User-specific LLM credentials.
- Audit logging.

These features will be introduced only when required.

### 10.10 Security Design Principles

1. Apply secure defaults.
2. Keep secrets outside source code.
3. Never expose cloud API keys to the frontend.
4. Validate uploaded PDFs.
5. Restrict access to application data.
6. Separate configuration from application logic.
7. Avoid logging sensitive information.
8. Use HTTPS in production.
9. Prepare for authentication and multi-user support.
10. Apply the principle of least privilege.
